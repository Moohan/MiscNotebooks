```python
import re, os

def tidy_syntax(filepath):
    #Store the syntax from the file in a variable
    syntax = open(filepath, "r").read()
    
    new_syntax = syntax

    #Replacments
    #Add spaces around operators
    new_syntax = re.sub(r"([:alnum:]|!|'|\")\s*?(?:(=)|(\+)|(>=?)|(<=?)|(\*{1}))\s*(.)", r"\1 \2\3\4\5\6 \7", new_syntax)  
    
    #Add spaces after commas (Works for up to 2; a,b,c -> a, b, c However will add double spaces which are cleaned later)
    new_syntax = re.sub(r"([:alnum:]){1}(?:,\s*(.*?){1}(,)\s*(.*?){1}|,\s*(.*?){1})", r"\1, \2\3\5 \4", new_syntax)
    
    #Remove double spaces except for start of line
    new_syntax = re.sub(r"(\S) {2,}(\S)", r"\1 \2", new_syntax)
    
    #Remove some redundant syntax (aggregate)
    new_syntax = re.sub(r"( *aggregate) outfile = \* mode = addvariables", r"\1", new_syntax)

    #Save file out
    new_file = os.path.join(os.path.split(filepath)[0], os.path.basename(filepath).replace(".sps", "_tidy.sps"))
    open(new_file, "w").write(new_syntax)
    print("File saved: " + new_file)
    
```

```python
#Example usage
import glob, os

dir = r"H:\Test"

for file in glob.iglob(os.path.join(dir, "*.sps")):
    print("Running on file: " + file)
    tidy_syntax(file)
```
