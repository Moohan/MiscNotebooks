```python
from flask import Flask, jsonify
from flask_cors import CORS
import pandas as pd

app = Flask(__name__)
cors = CORS(app) #Initializes Cross Origin Resource sharing for the application

csv = pd.read_csv('filter data.csv')

@app.route('/data')
def return_all_data():
    results = jsonify(csv.to_dict(orient='list'))
    return results

@app.route('/data/HB=<hb>&Sex=<sex>&Measure=<measure>')
def return_filtered_data(hb, sex, measure):
    #HB could be a comma-separated list:
    hb = hb.split(",")
    mask = (csv.HB.isin(hb)) & (csv.Sex==sex) & (csv.Measure==measure)
    csv_filtered = csv[mask]
    results = jsonify(csv_filtered.to_dict(orient='list'))
    return results
```
