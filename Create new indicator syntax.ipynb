```python
import re, os

def change_syntax(base_year):
    #Directory the 'base' syntax file is stored in
    dir = r"\\stats\irf\03-Integration-Indicators\01-Core-Suite\01-I12-13-20\June 18 syntax"
    os.chdir(dir)
    
    syntax_name_prefix = "01-I12-13-20-"
    base_syntax_name = syntax_name_prefix + base_year + ".sps"

    #Store the syntax from the file in a variable
    syntax = open(base_syntax_name).read()
    
    #Create a copy of the base syntax (just in case)
    open(syntax_name_prefix + base_year + "_COPY.sps", "w").write(syntax)
    
    years = ["1718"]
    
    quoted_base_year = "'" + base_year + "'"
    print("Base year is: " + quoted_base_year)
    for year in years:
        quoted_year = "'" + year + "'"
        print(quoted_year)
        
        #Replace 'FY'
        new_syntax = re.sub(quoted_base_year, quoted_year, syntax)
        
        #Replace some dates
        new_syntax = re.sub(str(2000 + int(base_year[0:2])) + r"([0-9]{4})", str(2000 + int(year[0:2])) + r"\1", new_syntax)
        new_syntax = re.sub(r"\(" + str(2000 + int(base_year[0:2])) + r",", r"(" + str(2000 + int(year[0:2])) + r",", new_syntax)
        #Replace some other dates
        new_syntax = re.sub(str(2000 + int(base_year[0:2]) + 1) + r"([0-9]{4})", str(2000 + int(year[0:2]) + 1) + r"\1", new_syntax)
        new_syntax = re.sub(r"\(" + str(2000 + int(base_year[0:2]) + 1) + r",", r"(" + str(2000 + int(year[0:2]) + 1) + r",", new_syntax)
        
        new_file = open(syntax_name_prefix + year + ".sps", "w")
        new_file.write(new_syntax)
        print(quoted_year + " saved to:")
        print(new_file.name)
        print()
        new_file.close()
```

```python
change_syntax(base_year='1314')
```
