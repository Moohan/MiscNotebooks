```python
def calc_check_digit(chi, factors = (10, 9, 8, 7, 6, 5, 4, 3, 2)):
    sum = 0
    for (f, n) in zip(factors, chi):
        sum += int(f) * int(n)

    chk =  sum % 11
    if chk == 0:
        return 5
    if chk == 1:
        return 0
    return 11 - chk

def check_digit_match(chi):
    if calc_check_digit(chi) == int(chi[-1]):
        return True
    else:
        return False
```

```python
james_chi = "1504915593"

check_digit_match(james_chi)
```

```python
def generate_chis():
    all_nums = [i for i in range(0, 9999999999) if check_digit_match(str(i))]
    print(all_nums[1])
    return all_nums
    
generate_chis()
```
