```python
def alterRenameSyntax(string):
    old, new = string.split("=")
    newSyntax = []
    for i in range(0, len(old.split())):
        newSyntax.append(old.split()[i] + " = " + new.split()[i])
    return alphabetizeList(newSyntax)

def alphabetizeList(syntax, printLines = True):
    if type(syntax) is str:
        syntax = [line.strip() for line in syntax.splitlines()]
    
    alphabetized = sorted(syntax)
    
    if printLines:
        for line in alphabetized:
            print(line)
            
    return alphabetized
```
